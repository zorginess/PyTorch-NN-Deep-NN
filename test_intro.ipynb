```python
device = torch.device(
    "mps" if torch.backends.mps.is_available() else "cpu"
)

print(device)
```

```python
x = torch.randn(1000, 1000, device=device)
y = torch.randn(1000, 1000, device=device)

z = x @ y
print(z.device)  # mps:0
```

```python
%%time
x = torch.tensor(torch.randn(10000, 10000, device=device))
y = torch.tensor(torch.randn(10000, 10000, device=device))

z = x @ y
print(z.device)  # mps:0
```

```python
%%time
x = torch.randn(10000, 10000, device='cpu')
y = torch.randn(10000, 10000, device='cpu')

z = x @ y
print(z.device)  # mps:0
```

```python
#create a random tensor and move it to the GPU if available
tensor = torch.tensor([1,2,3], device='cpu')

tensor, tensor.device
```
